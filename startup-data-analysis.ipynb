```python
import re

# Dictionary for known brand capitalizations & variation mappings
special_brands = {
    "1mg": "1mg",
    "20dresses": "20Dresses",
    "360ride": "360Ride",
    "3dexter": "3Dexter",
    "3hcare": "3HCare",
    "3rdflix": "3rdFlix",
    "48east": "48East",
    "4tigo": "4tigo",
    "5th vital": "5th Vital",
    "6degree": "6Degree",
    "75f": "75F",
    "91springboard": "91SpringBoard",
    "99games": "99Games",
    "99perhour": "99PerHour",
    "9stacks": "9Stacks",
    "byju's": "Byju's",
    "byjus": "Byju's",
    "byju": "Byju's",
    "ola": "Ola",
    "ola cabs": "Ola",
    "olacabs": "Ola",
    "oyo": "OYO Rooms",
    "oyo rooms": "OYO Rooms",
    "oyorooms": "OYO Rooms",
    "paytm": "Paytm",
    "paytm marketplace": "Paytm",
    "flipkart": "Flipkart",
    "swiggy": "Swiggy",
    "lendingkart": "Lendingkart",
    "lending kart": "Lendingkart",
    "urbanclap": "UrbanClap",
    "urban clap": "UrbanClap",
    "newsinshorts": "NewsInShorts",
    "news in shorts": "NewsInShorts",
    "phonepe": "PhonePe",
    "phone pe": "PhonePe",
    "pinelabs": "Pine Labs",
    "pine labs": "Pine Labs",
    "razorpay": "Razorpay",
    "razor pay": "Razorpay",
    "traveltriangle": "TravelTriangle",
    "travel triangle": "TravelTriangle",
    "urbanladder": "Urban Ladder",
    "urban ladder": "Urban Ladder",
    "wowexpress": "WoW Express",
    "wow express": "WoW Express",
    "zestmoney": "ZestMoney",
    "zest money": "ZestMoney",
    "i2i funding": "i2i Funding",
    "impactguru": "ImpactGuru",
    "impact guru": "ImpactGuru",
    "indialends": "IndiaLends",
    "india lends": "IndiaLends",
    "lendenclub": "LenDenClub",
    "lenden club": "LenDenClub",
    "letstransport": "LetsTransport",
    "lets transport": "LetsTransport",
    "loantap": "LoanTap",
    "loan tap": "LoanTap",
    "log9 materials": "Log9 Materials",
    "log 9 materials": "Log9 Materials",
    "milkbasket": "Milkbasket",
    "milk basket": "Milkbasket",
    "mswipe": "mSwipe",
    "ekincare": "eKincare",
    "mpaani": "mPaani",
    "zovi": "Zovi",
    "holachef": "HolaChef",
    "homelane": "HomeLane",
    "icanstay": "I Can Stay",
    "icustommadeit": "ICustomMadeIt",
    "inclov": "Inclov",
    "industrybuying": "IndustryBuying",
    "inkmonk": "InkMonk",
    "inmobi": "InMobi",
    "jolly food fellow": "Jolly Food Fellow",
    "jollyfoodfellow": "Jolly Food Fellow",
    "justride": "JustRide",
    "kaaryah": "KAARYAH",
    "kredx": "KredX",
    "leap india": "Leap India",
    "legalraasta": "LegalRaasta",
    "letsmd": "LetsMD",
    "livspace": "LivSpace",
    "loanzen": "LoanZen",
    "logicroots": "LogicRoots",
    "mamaearth": "Mamaearth",
    "mfine": "MFine",
    "mind your fleet": "Mind Your Fleet",
    "mindyourfleet": "MindYourFleet",
    "mobiefit": "MobieFit",
    "mobikwik": "MobiKwik",
    "mycity4kids": "MyCity4Kids",
    "nearbuy": "NearBuy",
    "nestaway": "NestAway",
    "netmeds": "NetMeds",
    "nightstay": "NightStay",
    "notesgen": "Notesgen",
    "ofbusiness": "OfBusiness",
    "paysense": "PaySense",
    "peppertap": "PepperTap",
    "pharmeasy": "PharmEasy",
    "pipabella": "Pipa Bella",
    "pipa bella": "Pipa Bella",
    "pitstop": "PitStop",
    "pi ventures": "Pi Ventures",
    "play your sport": "Play Your Sport",
    "policybazaar": "PolicyBazaar",
    "policy bazaar": "PolicyBazaar",
    "practo": "Practo",
    "pratilipi": "Pratilipi",
    "qwikspec": "QwikSpec",
    "raw pressery": "RAW Pressery",
    "rawpressery": "RAW Pressery",
    "reddoorz": "RedDoorz",
    "renewbuy": "RenewBuy",
    "renew buy": "RenewBuy",
    "rentomojo": "RentoMojo",
    "rollmafia": "RollMafia",
    "roll mafia": "Roll Mafia",
    "room on call": "Room On Call",
    "salebhai": "SaleBhai",
    "satvacart": "SatvaCart",
    "shadowfax": "ShadowFax",
    "shoekonnect": "ShoeKonnect",
    "shopclues": "ShopClues",
    "shopkirana": "ShopKirana",
    "shubh loans": "Shubh Loans",
    "shubhloans": "ShubhLoans",
    "sigtuple": "SigTuple",
    "smartcoin": "SmartCoin",
    "snapbizz": "SnapBizz",
    "spareshub": "SparesHub",
    "spares hub": "Spares Hub",
    "squadrun": "SquadRun",
    "stalk buy love": "StalkBuyLove",
    "stalkbuylove": "StalkBuyLove",
    "stayzilla": "Stayzilla",
    "stitchwood": "StitchWood",
    "styledotme": "StyleDotMe",
    "tails life": "TailsLife",
    "tailslife": "TailsLife",
    "tapchief": "TapChief",
    "teabox": "TeaBox",
    "tea box": "TeaBox",
    "testbook": "TestBook",
    "timesaverz": "TimeSaverz",
    "tinmen": "TinMen",
    "tonetag": "ToneTag",
    "tone tag": "ToneTag",
    "touchkin": "TouchKin",
    "townscript": "TownScript",
    "tripshelf": "TripShelf",
    "trip shelf": "TripShelf",
    "ue life sciences": "UE LifeSciences",
    "vanity cube": "VanityCube",
    "vanitycube": "VanityCube",
    "voxweb": "VoxWeb",
    "yourdost": "YourDOST",
    "zendrive": "ZenDrive",
    "zipgrid": "ZipGrid",
    "zoomcar": "ZoomCar"
}

def clean_startup(name):
    if pd.isna(name):
        return 'Unknown'
    s = str(name).strip()
    
    # 1. Handle byte escape representations of quotes, hyphens, non-breaking spaces
    s = re.sub(r'\\xe2\\x80\\x99|\\xe2\\x80\\x98', "'", s)
    s = re.sub(r'\\xe2\\x80\\x93|\\xe2\\x80\\x94', "-", s)
    s = re.sub(r'\\xc2\\xa0', " ", s)
    
    # 2. Remove byte escape sequences and whitespace escapes
    s = re.sub(r'\\x[a-fA-F0-9]{2}', '', s)
    s = re.sub(r'\\[nrt]', ' ', s)
    
    # 3. Clean quotes, backslashes, and replacement characters
    s = s.replace("\\\'", "'")
    s = s.replace('\\', '')
    s = s.replace('"', '')
    s = s.replace('#', '')
    s = s.replace("`", "'")
    s = s.replace("’", "'")
    s = s.replace("\ufffd", "'")
    
    # 4. Remove URL schemes, www, trailing paths, domain extensions
    s = re.sub(r'^https?://', '', s, flags=re.IGNORECASE)
    s = re.sub(r'^www\.', '', s, flags=re.IGNORECASE)
    s = re.sub(r'\/(.*)$', '', s)
    s = re.sub(r'\.(com|in|co|net|org)$', '', s, flags=re.IGNORECASE)
    
    # 5. Remove parenthetical details
    s = re.sub(r'\s*\([^)]*\)', '', s)
    
    # 6. Strip corporate suffixes
    s = re.sub(r'\s+(Pvt\.?\s*Ltd\.?|Private\s+Limited|Inc\.?|LLP\.?|Ltd\.?|Corp\.?)$', '', s, flags=re.IGNORECASE)
    s = re.sub(r'\s+Technologies\s+(Pvt\.?\s*Ltd\.?|Private\s+Limited)$', '', s, flags=re.IGNORECASE)
    
    # 7. Normalize spaces
    s = re.sub(r'\s+', ' ', s).strip()
    
    # 8. Check special brand dictionary
    s_lower = s.lower()
    if s_lower in special_brands:
        return special_brands[s_lower]
        
    # 9. Apply Proper Word/Title Casing if lower/upper
    if s.islower() or s.isupper():
        words = s.split(' ')
        cap_words = []
        for w in words:
            if not w:
                continue
            parts = w.split('-')
            cap_parts = []
            for p in parts:
                if "'" in p:
                    sub = p.split("'")
                    cap_p = sub[0].capitalize() + "'" + (sub[1].lower() if len(sub) > 1 else '')
                else:
                    cap_p = p.capitalize()
                cap_parts.append(cap_p)
            cap_words.append('-'.join(cap_parts))
        s = ' '.join(cap_words)
        
    return s

funding['Startup'] = funding['Startup'].apply(clean_startup)
funding['Startup'] = funding['Startup'].fillna('Unknown')
```
