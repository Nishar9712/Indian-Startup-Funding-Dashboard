## Handle Investment Amount($) colomn

```python
funding['Investment Amount($)'] = funding['Investment Amount($)'].fillna('unknown')
funding['Investment Amount($)']
```

```python
# Remove commas(,)
funding['Investment Amount($)'].astype('string') + ','
funding['Investment Amount($)'] = funding['Investment Amount($)'].astype('string').str.replace(',', '', regex=False)
```

```python
funding['Investment Amount($)']
```

```python
funding['Investment Amount($)'][funding['Investment Amount($)'].str.contains(r'[A-Za-z]', regex=True, na=False)].value_counts()
```

```python
funding['Investment Amount($)'] = funding['Investment Amount($)'].str.replace(r'\\xc2\\xa0','')
```

```python
funding['Investment Amount($)'] = pd.to_numeric(funding['Investment Amount($)'], errors='coerce')

funding['Investment Amount($)'].isna().sum()
```

```python
check = funding.groupby('Startup').agg(
    Total_Rows=('Investment Amount($)', 'size'),
    NaN_Amount=('Investment Amount($)', 'count')
)

check['NaN_Amount'] = check['Total_Rows'] - check['NaN_Amount']

nan_values = check[check['Total_Rows'] == 1]

nan_values['NaN_Amount'].sum()
```

```python
fill = funding.groupby('Startup')['Investment Amount($)'].transform('mean')
funding['Investment Amount($)'] = funding['Investment Amount($)'].fillna(fill)
funding['Investment Amount($)'] = funding['Investment Amount($)'].fillna(funding['Investment Amount($)'].mean())
```

```python
funding['Investment Amount($)'] = funding['Investment Amount($)'].astype('string')
funding['Investment Amount($)'][funding['Investment Amount($)'].str.contains(r'[A-Za-z]', regex=True, na=False)].value_counts()
```

```python
funding['Investment Amount($)'] = funding['Investment Amount($)'].astype('float')
```

## Handle Investment Type colomn

```python
funding['Investment Type'].isna().sum()
```

```python
# funding['Investment Type'].value_counts()
```

```python
funding['Investment Type'].unique()
```

```python
funding['Investment Type'] = funding['Investment Type'].str.replace(r'Venture - Series Unknown','Venture - Series')
funding['Investment Type'] = funding['Investment Type'].str.replace(r'Seed/ Angel Funding','Seed / Angel Funding')
funding['Investment Type'] = funding['Investment Type'].str.replace(r'Seed / Angle Funding','Seed / Angel Funding')
funding['Investment Type'] = funding['Investment Type'].str.replace(r'Seed/Angel Funding','Seed / Angel Funding')
funding['Investment Type'] = funding['Investment Type'].str.replace(r'Angel / Seed Funding','Seed / Angel Funding')
funding['Investment Type'] = funding['Investment Type'].str.replace(r'Seed\\nFunding','Seed Funding')
funding['Investment Type'] = funding['Investment Type'].str.replace(r'Private\\nEquity','Private Equity')
```

```python
fill = funding.groupby('Startup')['Investment Type'].transform(
    lambda x: x.mode().iloc[0] if not x.mode().empty else 'Not Know'
)

funding['Investment Type'] = funding['Investment Type'].fillna(fill)
```
